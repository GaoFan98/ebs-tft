# EBS native-resolution TFT/DeepLOB local pilot

This is a bounded engineering smoke test, not publication evidence. It calls the reusable project use case and then inspects its generated artifacts.

```python
from pathlib import Path

import polars as pl

from ebs_tft.application.usecases import pilot

repository = Path.cwd()
if repository.name == "notebooks":
    repository = repository.parent
config_path = repository / "notebooks" / "pilot_smoke.yaml"
specification = pilot.load_specification(path=config_path)
specification
```

```python
result = pilot.run(specification=specification)
result
```

```python
target_balance = pl.read_csv(result.output_dir / "target_balance.csv")
target_balance
```

```python
metrics = pl.read_csv(result.metrics_path)
metrics.sort(["horizon_steps", "training_mode", "model", "depth", "seed"])
```

```python
predictions = pl.read_parquet(result.predictions_path)
group_keys = ["horizon_steps", "training_mode", "model", "depth", "seed"]
predictions.group_by(group_keys).len().sort(group_keys)
```

Interpret only whether the pipeline runs, targets align, and models produce probabilities. Each modeled horizon is a separate fit over the same untouched 100 ms state table. Do not interpret this bounded pilot as evidence that one model or depth is superior.
